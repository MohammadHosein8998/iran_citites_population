```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import ScalarFormatter
```

## adding data

## Building plots

### plot

```python
plt.figure(figsize=(10, 10))
plt.plot(cities, population, lw=1.5, linestyle="dashed", marker="P", color="black",
         label="population in top 10 Iran cities")

formatter = ScalarFormatter()
formatter.set_scientific(False)
plt.gca().yaxis.set_major_formatter(formatter)

plt.xlabel("cities")
plt.ylabel("population")
plt.legend(loc="upper right")
plt.show()
```

### scatter

```python
plt.figure(figsize=(10, 10))
plt.scatter(cities, population, lw=1.5, marker="v", color="red", label="population in top 10 Iran cities")

formatter.set_scientific(False)
plt.gca().get_yaxis().set_major_formatter(formatter)

plt.xlabel("cities")
plt.ylabel("population")
plt.legend(loc="upper right")
plt.show()
```

### bar

```python
plt.figure(figsize=(10, 10))
plt.bar(cities, population, linestyle="dashed", color="blue", label="population in top 10 Iran cities")

formatter.set_scientific(False)
plt.gca().get_yaxis().set_major_formatter(formatter)

plt.xlabel("cities")
plt.ylabel("population")
plt.legend(loc="upper right")
plt.show()
```

### barh

```python
plt.figure(figsize=(10, 10))
plt.barh(cities, population, linestyle="dashed", color="blue", label="population in top 10 Iran cities")

formatter.set_scientific(False)
plt.gca().get_xaxis().set_major_formatter(formatter)

# plt.ylabel("cities")
# plt.xlabel("population")
plt.legend(loc="upper right")
plt.show()
```

### pie

```python
plt.figure(figsize=(10, 10))
plt.pie(population, labels=cities, autopct="%1.1f%%", startangle=90)

plt.title("Population Distribution by City")
# formatter.set_scientific(False)
# plt.gca().get_yaxis().set_major_formatter(formatter)

# plt.xlabel("cities")
# plt.ylabel("population")
# plt.legend(loc="upper right")
plt.show()
```

```python
import plotly.express as px

# داده‌های شهرها، جمعیت و مختصات جغرافیایی
data = {
    "City": cities,
    "Population": population,
    "Lat": latitudes,
    "Lon": longitudes
}

# رسم نقشه حبابی روی نقشه جهان/منطقه
fig = px.scatter_geo(
    data,
    lat="Lat",
    lon="Lon",
    size="Population",  # اندازه دایره متناسب با جمعیت
    hover_name="City",  # نمایش نام شهر با هاور کردن ماوس
    size_max=50,  # حداکثر اندازه دایره‌ها
    projection="mercator",  # نوع تصویرسازی نقشه
    title="Iran Cities Population Map"
)

# محدود کردن کادر نقشه روی محدوده ایران
fig.update_geos(
    visible=True,
    resolution=50,
    scope="asia",
    showcountries=True,
    countrycolor="gray",
    lonaxis_range=[44, 63],  # محدودیت طول جغرافیایی ایران
    lataxis_range=[24, 40]  # محدودیت عرض جغرافیایی ایران
)

# fig.show()
# ذخیره خروجی به صورت فایل HTML در کنار پروژه
fig.write_html("iran_map.html", auto_open=True)
```
